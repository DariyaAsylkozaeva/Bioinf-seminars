task 1
Напишите функцию, которая будет возвращать индекс минимального элемента в списке. Если их несколько, то вернет все минимальные положения.

```python
def min_index(input_list: list): # Объявляем функцию, которая принимает на вход список
     # Проверка 1: входной параметр — список
     assert isinstance(input_list, list), "Входной параметр не является списком"
     # Проверка 2: список не пустой
     assert len(input_list) > 0, "Входной параметр является пустым списком"
     # Предполагаем, что первый элемент — это минимум
     current_min = input_list[0]
     # Создаём список индексов минимальных элементов, куда добавляем индекс 0, т.к. считаем первый элемент минимальным
     index = [0] 
     # Проходим по всем элементам списка
     # enumerate() возвращает пары (индекс, значение) для каждого элемента
     for i, value in enumerate(input_list):
          if value < current_min:  # Если нашли элемент меньше текущего минимума
                current_min = value  # Обновляем минимум 
                index = [i]  # Обновляем список индексов минимальных элементов
          elif value == current_min: # Если нашли элемент равный текущему минимуму
                      index.append(i)  # Добавляем его индекс в список
     return index   # Возвращаем список индексов минимальных элементов
# Пример вызова функции
example_list = [5,-4,4556,123,44,-5,0,324]
min_value = min_index(example_list)
print(min_value)             
```

Напишите функцию, которая возвращает отсортированный методом пузырька список. 

```python

def bubble_sort(input_list): # Объявляем функцию, которая принимает список для сортировки
    n = len(input_list) # Сохраняем длину списка в переменную n
    # Внешний цикл: количество проходов по списку
    # Для списка из n элементов достаточно n-1 проходов
    for i in range(n - 1):
        # Внутренний цикл: сравнение соседних элементов
        # range(n - 1 - i) — с каждым проходом проверяем на 1 элемент меньше
        for j in range(n - 1 - i):
            # Сравниваем текущий элемент с следующим
            if input_list[j] > input_list[j + 1]:
                # Меняем элементы местами 
                input_list[j], input_list[j + 1] = input_list[j + 1], input_list[j]
    return input_list # Возвращаем отсортированный список
# Пример вызова функции
input_list=[5,7,3,8,9,1]
input_list=bubble_sort(input_list)
print(input_list)


def decorator_function(func):
    def wrapper(*args):  # Функция обертка будет принимать позиционные аргументы для func
        print('='*60)
        print('Функция-обёртка!')
        print(f'Оборачиваемая функция: {func}')
        print('Выполняем обёрнутую функцию...')
        print('-'*60)
        # Вызываем функцию с переданными аргументами
        result = func(*args) # Сохраняем то, что возвращает наша func в result
        print('-'*60)
        print(f'Результат: {result}')
        print('Выходим из обёртки')
        return result # Возвращаем результат
    return wrapper

@decorator_function
def bubble_sort(input_list): # Объявляем функцию, которая принимает список для сортировки
    n = len(input_list) # Сохраняем длину списка в переменную n
    # Внешний цикл: количество проходов по списку
    # Для списка из n элементов достаточно n-1 проходов
    for i in range(n - 1):
        # Внутренний цикл: сравнение соседних элементов
        # range(n - 1 - i) — с каждым проходом проверяем на 1 элемент меньше
        for j in range(n - 1 - i):
            # Сравниваем текущий элемент с следующим
            if input_list[j] > input_list[j + 1]:
                # Меняем элементы местами 
                input_list[j], input_list[j + 1] = input_list[j + 1], input_list[j]
    return input_list # Возвращаем отсортированный список
# Пример вызова функции
input_list=[5,7,3,8,9,1]
input_list=bubble_sort(input_list)
print(input_list)
    
```

Напишите функцию-генератор случайной нуклеотидной последовательности заданной пользователем длины.

```python
import random # Для генераций случайных чисел (в нашем случае нуклеотидов)
def sequence(length): # Объявляем функцию, параметр — длина последовательности
    nucleotides = ['A', 'T', 'G', 'C'] # Создаём список из 4 нуклеотидов ДНК
    result=[] # Создаём пустой список, куда будем собирать случайные нуклеотиды
    for i in range(length): # Цикл повторяется 'length' раз
        # random.choice(nucleotides) выбирает один случайный элемент из списка
        # append() добавляет этот элемент в конец списка result
        result.append(random.choice(nucleotides))
    return result # Возвращает готовый список 
# Пример вызова функции
dna=sequence(10)
print(dna)
```

Task 4
Реализуйте в виде функции игру “Угадай число”. Пользователю дается 10 попыток угадать заданное случайное число в диапазоне от 1 до 100. При каждой попытке программа должна оповещать игрока, загаданное число больше, меньше или равно введенного.

```python
import random # Для генерации случайных чисел
def guess_the_number(): # Объявляем функцию 
    secret_number = random.randint(1, 100) # Генерируем случайное целое число от 1 до 100
    max_attempts = 10 # Задаём максимальное количество попыток
    print("Добро пожаловать в игру 'Угадай число'")
    print("Я загадал число от 1 до 100. У вас 10 попыток")
    for attempt in range(1, max_attempts + 1): # Цикл на 10 попыток
        # Запрашиваем число у игрока
        # input() возвращает строку, int() превращает её в число
        # f-строка подставляет номер текущей попытки
        user_number = int(input(f"Попытка {attempt}/{max_attempts}. Введите число: "))
        # Если угадал
        if user_number == secret_number:
            print(f"Вы угадали число {secret_number} за {attempt} попыток")
            return # Завершает функцию
        # Если загаданное число больше
        elif user_number < secret_number:
            print("Загаданное число больше вашего")
        # Если загаданное число меньше
        else:
            print("Загаданное число меньше вашего")
    # Если за 10 попыток ни разу не угадал
    print(f"Попытки закончились. Загаданное число было: {secret_number}")

# Запуск игры
guess_the_number()
```

Fibonacci Numbers

```python
# Каждое число — сумма двух предыдущих
def fibonacci(n): 
    if n == 0:
        return 0
    if n == 1:
        return 1
    a,b=0,1
    for i in range(2,n+1):
        # Множественное присваивание
        a,b=b,a+b
    return b
fibonacci(20)
```

написать функцию, которая для любого входного графа такого формата выводит степени всех его вершин.

```python
def degree_array(graph_description:list, edges_list: list): #принимает список состоящий из кол-ва вершин и ребер и список ребер
    assert isinstance(graph_description, list) and len(graph_description) == 2, "Входной параметр не cоответсвует требованию"

    vertices, edges = graph_description
    result_degrees = [0 for _ in range(vertices)]  #генерируем список из 0 по кол-ву вершин 
    for i in edges_list:
        v1, v2 = [int(k) for k in i.split()] #получаем две вершины, соединенные ребром
        result_degrees[v1-1] += 1 #добавляем 1 к степени нужной вершины в результирующем списке
        result_degrees[v2-1] += 1
    return result_degrees


edges = [
    '1 2',
    '1 5',
    '2 3',
    '2 6',
    '3 4',
    '3 7',
    '4 8',
    '5 6',
    '6 7',
    '7 8'
]
print(degree_array([8, 10], edges))
```

Insertion Sort 
Напиши программу, которая получает массив чисел, сортирует его методом insertion sort и считает, сколько обменов элементов произошло в процессе сортировки.

```python
def insertion_sort(input_list):
    swaps=0 # Счетчик перестановок
    for i in range(1,len(input_list)):
        current=input_list[i]  # Сохраняем текущий элемент в отдельную переменную
        j=i-1
        while j >= 0 and input_list[j] > current:
            # Сдвигаем элемент input_list[j] на одну позицию вправо
            # Это создает "пустое" место для current
            input_list[j+1]=input_list[j]
            swaps += 1
            # Уменьшаем j на 1, чтобы проверить следующий элемент слева
            # Мы двигаемся влево, потому что нужно сравнить current со всеми предыдущими элементами
            j -= 1
            input_list[j+1]=current
    return input_list, swaps 
input_list=[1,5,3,7,2,9,8]
sorted_list=insertion_sort(input_list)
print(sorted_list)
```

Complementing a Strand of DNA

```python
def reverse_complement(dna):
    replacements = {'A': 'T', 'T': 'A', 'C': 'G', 'G': 'C'}
    complemented = ''
    for nucleotide in dna:
        complemented += replacements[nucleotide]
    return complemented[::-1]

dna = 'AAAACCCGGT'
result = reverse_complement(dna)
print(result)

    
   
```

Rabbits and Recurrence Relations

```python
def rabbit_pairs(n, k):
    # Формула:
    # F(1) = 1
    # F(2) = 1
    # F(n) = F(n-1) + k * F(n-2)  (для n > 2)
    if n == 1:
        return 1
    if n == 2:
        return 1
    ready = 1  # F(n-2)
    notready = 1       # F(n-1)
    for month in range(3, n + 1):
        current = notready + k * ready
        ready = notready
        notready = current
    return current
rabbits=rabbit_pairs(5, 3)
print(rabbits)
```
